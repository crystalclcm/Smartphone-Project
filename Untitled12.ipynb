```python
import pandas as pd
```

```python
df = pd.read_csv('/smartphone_dataset.csv')
```

```python
print(df.shape)
print(df.head())
print(df.info())
```

```python

# Normalize brand and model names
df['brand_name'] = df['brand_name'].str.lower()
df['model'] = df['model'].str.lower()

# Check for missing values
print(df.isnull().sum())
```

```python
df = df.dropna()
```

```python
df['rating'] = df['rating'].fillna(df['rating'].mean())
```

```python
df['processor_brand'] = df['processor_brand'].fillna('Unknown')
```

```python
df['has_5g'] = df['has_5g'].fillna(False)
```

```python
df = df.drop_duplicates()
```

```python
print(df.dtypes)
```

```python
df['price'] = pd.to_numeric(df['price'], errors='coerce')
```

```python
print(df.info())
print(df.describe())
```

```python
print(df.columns.tolist())
```

```python
display(df.head())
```

```python
df = df.dropna(subset=['price','rating'])
```

```python
brand_stats = df.groupby('brand_name').agg({'price':'mean','rating':'mean'}).reset_index()
```
